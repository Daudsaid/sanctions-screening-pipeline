```python
from pathlib import Path

nat = q(r"""
select raw->>'nationality' as nationality, count(*) as n
from raw_ch_officer
where raw->>'officer_role' not like 'corporate%' and raw->>'nationality' is not null
group by 1 order by 2 desc
""")

nat = nat.with_columns(
    (nat["n"].cum_sum() / nat["n"].sum()).alias("cumulative_share")
)
print("values needed for 95%:", (nat["cumulative_share"] < 0.95).sum() + 1)
print("values needed for 99%:", (nat["cumulative_share"] < 0.99).sum() + 1)

nat.write_csv("nationality_spellings.csv")
print("wrote", Path("nationality_spellings.csv").resolve())
```
